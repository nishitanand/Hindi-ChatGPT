```python
import datasets
from datasets import Dataset, Features, Audio, load_dataset
# audio = ["input_audio/record4_novel_summary.wav"] * 10
# audio = ["input_audio/file.wav"] * 10
audio = ["input_audio/input_query.wav"] * 10
features = Features({"audio": Audio(sampling_rate=16000)})
ds = Dataset.from_dict({"audio": audio}, features=features)
ds = ds.with_format("torch")
print(ds[0]["audio"]["array"])
print("Sampling Rate ", ds[0]["audio"]["sampling_rate"])
print("Audio imported successfully")
```

```python
hindi_input_audio=ds[0]["audio"]["array"]

from transformers import Wav2Vec2ForCTC, AutoProcessor
import torch


device = "cuda:0" if torch.cuda.is_available() else "cpu"
print(device)


model_id = "facebook/mms-1b-all"

processor = AutoProcessor.from_pretrained(model_id)
model = Wav2Vec2ForCTC.from_pretrained(model_id)


processor.tokenizer.set_target_lang("hin")
model.load_adapter("hin")


inputs = processor(hindi_input_audio, sampling_rate=16_000, return_tensors="pt")

with torch.no_grad():
    outputs = model(**inputs).logits

ids = torch.argmax(outputs, dim=-1)[0]
transcription = processor.decode(ids)

print("Query in Hindi: ", transcription)


hindi_input_text=transcription
```

```python
from transformers import AutoModelForSeq2SeqLM, AutoTokenizer

tokenizer = AutoTokenizer.from_pretrained(
    "facebook/nllb-200-distilled-600M", src_lang="hin_Deva"
)
model = AutoModelForSeq2SeqLM.from_pretrained("facebook/nllb-200-distilled-600M")

inputs = tokenizer(hindi_input_text, return_tensors="pt")
translated_tokens = model.generate(
    **inputs, forced_bos_token_id=tokenizer.lang_code_to_id["eng_Latn"], max_length=3000
)
english_input_text=tokenizer.batch_decode(translated_tokens, skip_special_tokens=True)[0]
print("Query Translated to English: ", english_input_text)
```

```python

from transformers import AutoModelForSeq2SeqLM, AutoTokenizer

tokenizer = AutoTokenizer.from_pretrained(
    "facebook/nllb-200-distilled-600M", src_lang="eng_Latn"
)
model = AutoModelForSeq2SeqLM.from_pretrained("facebook/nllb-200-distilled-600M")

inputs = tokenizer(english_output_text, return_tensors="pt")

translated_tokens = model.generate(
    **inputs, forced_bos_token_id=tokenizer.lang_code_to_id["hin_Deva"], max_length=3000
)
hindi_output_text=tokenizer.batch_decode(translated_tokens, skip_special_tokens=True)[0]
print("Model Response Translated to Hindi: ", hindi_output_text)
```

```python
from transformers import VitsModel, AutoTokenizer
import torch

model = VitsModel.from_pretrained("facebook/mms-tts-hin")
tokenizer = AutoTokenizer.from_pretrained("facebook/mms-tts-hin")

text = hindi_output_text
inputs = tokenizer(text, return_tensors="pt")

with torch.no_grad():
    output = model(**inputs).waveform



import scipy

scipy.io.wavfile.write("output_audio/response.wav", rate=model.config.sampling_rate, data=output.float().numpy().T)


print("Hindi Audio Response Successfully Generated")

from IPython.display import Audio

Audio(output.numpy(), rate=model.config.sampling_rate)




# (llama2_quant) nishit@nishit-ubuntu:~/Downloads/llama2-cpp/llama.cpp$ make -j && ./main -m models/llama-2-7b-chat.Q4_K_M.gguf -p "Building a website can be done in 10 simple steps:\nStep 1:" -n 400 -e

# (llama2_quant) nishit@nishit-ubuntu:~/Downloads/llama2-cpp/llama.cpp$ make -j && ./main -ngl 32 -m models/llama-2-7b-chat.Q4_K_M.gguf --color -c 4096 --temp 0.7 --repeat_penalty 1.1 -n -1 -p "[INST] <<SYS>>\nBuilding a website can be done in 10 simple steps:\nStep 1:\n<</SYS>>\n{prompt}[/INST]"
```
